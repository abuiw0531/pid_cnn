```python
import serial, struct, math, time, threading
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
import random
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, GlobalAveragePooling1D, Dense

# ==========================================
# 1. 序列埠與全域狀態初始化
# ==========================================
try:
    USB_po = serial.Serial('COM4', 115200, timeout=1)  # 序列埠初始化
    print("序列埠 COM4 開啟成功！")
except Exception as e:
    print(f"序列埠開啟失敗或未連接: {e} (如無硬體可進行離線除錯)")
    USB_po = None

STATE = 1                # 程式狀態 (1:等待設定目標  2:發送參數  3:收集波形  4:波形分析與CNN學習)
command_ID = 0           # 發送給 Arduino 端的命令 ID
Arduino_command_ID = 0   # Arduino 當前確認的命令 ID
target_position = 0.0    # 當前目標位置 (cm)
last_target_position = 0.0 # 前一次目標位置 (cm)
target = 0               # 發送給 Arduino 的編碼器脈衝目標值

# 多執行緒安全鎖與資料緩衝區
data_lock = threading.Lock()
x, y = [], []            # 繪圖用波形資料 (時間, 位置)
CNN_input_1 = []         # 時間點序列 (s)
CNN_input_2 = []         # 目標值序列 (cm)
CNN_input_3 = []         # 實際位置序列 (cm)
CNN_input_4 = []         # 誤差序列 (cm)

# ==========================================
# 2. PID 與超參數管理
# ==========================================
eps = 1e-9
best_loss_map = {}
best_K_map = {}

# 分離不同目標階躍幅度的初始 K 參數 (Kpp, Kpd, Kvp, Kvi)
current_K_map = {
    "small_step": [0.001, 0.0002, 0.5, 0.01],
    "medium_step": [0.001, 0.0002, 0.5, 0.01],
    "large_step": [0.001, 0.0002, 0.5, 0.01]
}
current_noise_scale_map = {
    "small_step": 0.02,
    "medium_step": 0.02,
    "large_step": 0.02
}

# 經驗回放池 (Experience Replay)
replay_buffer = []  
MAX_REPLAY_SIZE = 50
state_flag = 0
FIXED_LEN = 200          # 1D-CNN 輸入時序固定長度


#### 讀取 / 切換使用者輸入目標 #############################################################
def Wait_input():
    global STATE, command_ID, state_flag
    global last_target_position, target_position, target

    last_target_position = target_position
    
    # 在 20 cm 與 0.5 cm 間交替測試階躍響應
    if state_flag == 0:
        state_flag = 1
        target_position = 20.0
    else:
        state_flag = 0
        target_position = 0.5
        
    target_position = float(np.clip(target_position, 0.0, 27.0))
    target_angle = -90.0 * target_position

    if target_angle >= 0:
        target = math.floor((target_angle * 4096 / 360) + 0.5)
    else:
        target = math.ceil((target_angle * 4096 / 360) - 0.5)

    print(f"\n[State 1] 切換目標位置: {target_position:.2f} cm (編碼器脈衝: {target})")
    command_ID += 1
    STATE = 2
    print('[State 1 -> 2] 準備發送控制參數至 Arduino...')


#### 序列埠接收執行緒 #####################################################################
def Receive():
    global STATE, Arduino_command_ID

    while True:
        if USB_po is None or not USB_po.is_open:
            time.sleep(0.5)
            continue

        try:
            P = USB_po.read(1)
            if len(P) != 1:
                continue

            data = struct.unpack('<B', P)
            if data[0] == 1:
                P = USB_po.read(5)
                if len(P) != 5:
                    continue
                data = struct.unpack('<IB', P)
                if data[1] == 2:
                    num1 = data[0]

                    P = USB_po.read(5)
                    if len(P) != 5:
                        continue
                    data = struct.unpack('<IB', P)
                    if data[1] == 3:
                        num2 = data[0]

                        P = USB_po.read(5)
                        if len(P) != 5:
                            continue
                        data = struct.unpack('<iB', P)
                        if data[1] == 6:
                            Arduino_command_ID = num1
                            if STATE == 3:
                                Data_collect(num2, data[0])
        except Exception as e:
            print(f"Receive 異常: {e}")
            time.sleep(0.1)


#### 序列埠發送執行緒 #####################################################################
def Transmit():
    global STATE, command_ID, Arduino_command_ID, target, target_position

    while True:
        time.sleep(0.05)
        if STATE == 2 and USB_po is not None and USB_po.is_open:
            try:
                if Arduino_command_ID != command_ID:
                    target_key = get_target_key(target_position)
                    current_K = current_K_map[target_key]
                    P = struct.pack(
                        '<BIBiBfBfBfBfB',
                        0x01, command_ID, 0x02, target,
                        0x03, float(current_K[0]),
                        0x04, float(current_K[1]),
                        0x05, float(current_K[2]),
                        0x06, float(current_K[3]),
                        0x07
                    )
                    USB_po.write(P)
                    print(f"  [Transmit] 命令 {command_ID} 已發送 (K: {[round(k, 5) for k in current_K]})")
                else:
                    STATE = 3
                    print('  [State 2 -> 3] Arduino 已確認，開始收集波形資料...')
            except Exception as e:
                print(f"Transmit 異常: {e}")


#### 收集即時波形資料 (執行緒安全) #########################################################
def Data_collect(Arduino_moment, Arduino_encoder):
    global STATE, Arduino_command_ID, target_position

    sample_time = Arduino_moment / 1000000.0
    angle = Arduino_encoder * 360.0 / 4096.0
    position = -angle / 90.0

    with data_lock:
        x.append(sample_time)
        CNN_input_1.append(sample_time)
        y.append(position)
        CNN_input_3.append(position)
        CNN_input_2.append(target_position)
        CNN_input_4.append(target_position - position)

    if sample_time >= 10.0 and STATE == 3:
        STATE = 4
        print(f"  [State 3 -> 4] 已收集滿 10 秒波形 (共 {len(x)} 點)，開始進行 CNN 評估與學習...")


#### 繪製波形圖 (記憶體安全釋放) ###########################################################
def Draw(t_data, y_data, target_val, y_ref_data=None):
    fig, ax = plt.subplots(figsize=(12, 3.5))
    ax.axhline(y=target_val, color='r', linestyle='--', label=f'Target ({target_val:.2f} cm)')
    if y_ref_data is not None and len(y_ref_data) == len(t_data):
        ax.plot(t_data, y_ref_data, color='green', linestyle=':', label='2nd-order Ideal Ref')
    ax.plot(t_data, y_data, color='black', marker='.', markersize=2, label='Actual Position')
    
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Position (cm)")
    ax.set_title(f"Step Response to {target_val:.2f} cm")
    ax.grid(True, linestyle=':')
    ax.legend(loc='best')
    plt.tight_layout()
    plt.show()
    plt.close(fig)  # 及時釋放避免 Matplotlib 記憶體洩漏


#### 計算 ITAE 誤差積分 ###################################################################
def compute_ITAE(t, e):
    t = np.array(t, dtype=np.float32)
    e = np.array(e, dtype=np.float32)
    if len(t) < 2:
        return 0.0

    dt = np.diff(t)
    dt = np.append(dt, dt[-1])
    return float(np.sum(t * np.abs(e) * dt))


#### 計算 % Overshoot 超調量 (支援雙向階躍) ################################################
def Get_Overshoot(y_arr, y0, target_val):
    if len(y_arr) == 0:
        return 0.0
    displacement = target_val - y0
    if abs(displacement) < 1e-5:
        return 0.0

    if displacement > 0:
        max_val = float(np.max(y_arr))
        return max((max_val - target_val) / displacement, 0.0)
    else:
        min_val = float(np.min(y_arr))
        return max((target_val - min_val) / abs(displacement), 0.0)


#### 雙向階躍二階系統理想響應 (支援任意初始位置 y0) ########################################
def generate_second_order_response(t, zeta, wn, y0, target_val):
    t = np.array(t, dtype=np.float32)
    t_shifted = t - t[0] if len(t) > 0 else t

    zeta = np.clip(zeta, 0.01, 0.99)
    wd = wn * np.sqrt(1.0 - zeta ** 2)
    phi = np.arccos(zeta)

    # 標準二階步階歸一化過渡函數
    step_norm = 1.0 - (1.0 / np.sqrt(1.0 - zeta ** 2)) * np.exp(-zeta * wn * t_shifted) * np.sin(wd * t_shifted + phi)
    
    # 支援 y0 到 target_val 的平滑過渡
    y_ref = y0 + (target_val - y0) * step_norm
    return y_ref


#### 波形時序特徵重採樣 (轉為固定長度 1D-CNN 輸入) #########################################
def resample_waveforms(t, target_arr, pos_arr, err_arr, fixed_len=200):
    t = np.array(t, dtype=np.float32)
    t_norm = np.linspace(t[0], t[-1], fixed_len)
    
    target_res = np.interp(t_norm, t, target_arr)
    pos_res = np.interp(t_norm, t, pos_arr)
    err_res = np.interp(t_norm, t, err_arr)
    time_res = np.linspace(0.0, 1.0, fixed_len)  # 時間標準化至 0~1

    # 特徵維度: (fixed_len, 4)
    feature = np.stack([time_res, target_res, pos_res, err_res], axis=-1).astype(np.float32)
    return feature


def get_target_key(position):
    position = float(position)
    if position < 2.0:
        return "small_step"
    if position < 10.0:
        return "medium_step"
    return "large_step"


# ==========================================
# 3. 1D-CNN 網路架構與優化器
# ==========================================
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-4)

# 使用 1D-CNN 提取完整的時序動態響應特徵
model = Sequential([
    Conv1D(32, kernel_size=5, activation='relu', padding='same', input_shape=(FIXED_LEN, 4)),
    MaxPooling1D(pool_size=2),
    Conv1D(64, kernel_size=5, activation='relu', padding='same'),
    MaxPooling1D(pool_size=2),
    Conv1D(128, kernel_size=3, activation='relu', padding='same'),
    GlobalAveragePooling1D(),
    Dense(64, activation='relu'),
    Dense(4, activation='relu')   # 輸出 4 個推薦的 K 參數值 (Kpp, Kpd, Kvp, Kvi)
])

# 啟動通訊執行緒
R = threading.Thread(target=Receive, daemon=True)
T = threading.Thread(target=Transmit, daemon=True)
R.start()
T.start()


# ==========================================
# 4. 主控迴圈 (自適應學習與探索)
# ==========================================
while True:
    if STATE == 1:
        Wait_input()

    if STATE == 4:
        # 安全複製並提取這 10 秒的採樣波形
        with data_lock:
            raw_t = list(CNN_input_1)
            raw_target = list(CNN_input_2)
            raw_pos = list(CNN_input_3)
            raw_err = list(CNN_input_4)
            # 清空緩衝區
            x.clear()
            y.clear()
            CNN_input_1.clear()
            CNN_input_2.clear()
            CNN_input_3.clear()
            CNN_input_4.clear()

        N = len(raw_t)
        if N < 20:
            print(f"[Warn] 收集到的數據點過少 ({N} 點)，跳過本輪訓練。")
            STATE = 1
            continue

        t_arr = np.array(raw_t, dtype=np.float32)
        target_arr = np.array(raw_target, dtype=np.float32)
        pos_arr = np.array(raw_pos, dtype=np.float32)
        err_arr = np.array(raw_err, dtype=np.float32)

        # 取得起始位置與目標位置
        y0 = float(pos_arr[0])
        current_target = float(target_position)

        # 二階理想響應標準 (設定阻尼比與自然頻率)
        OS_spec = 0.02
        zeta = -np.log(OS_spec) / np.sqrt(np.pi ** 2 + (np.log(OS_spec)) ** 2)
        Ts_spec = 4.0
        wn = 4.0 / (zeta * Ts_spec)

        y_ref = generate_second_order_response(t_arr, zeta, wn, y0, current_target)

        # 繪製當前波形與理想參考曲線
        Draw(t_arr, pos_arr, current_target, y_ref)

        # 計算綜合控制品質評分 (Tracking Loss + ITAE + Overshoot)
        tracking_loss = np.mean((pos_arr - y_ref) ** 2)
        itae_loss = compute_ITAE(t_arr, err_arr)
        overshoot_val = Get_Overshoot(pos_arr, y0, current_target)

        # 歸一化尺度常數
        delta_scale = max(abs(current_target - y0), 1.0)
        t_max = max(np.max(t_arr), 1.0)

        norm_tracking = tracking_loss / (delta_scale ** 2)
        norm_itae = itae_loss / (delta_scale * (t_max ** 2) * 0.5)
        norm_overshoot = overshoot_val

        loss_wave = float(norm_tracking + 0.4 * norm_itae + 0.2 * norm_overshoot)

        print(f"\n--- [評估結果] 目標: {current_target:.2f} cm (初始: {y0:.2f} cm) ---")
        print(f"波形綜合 Loss: {loss_wave:.6f} (Tracking: {norm_tracking:.4f}, ITAE: {norm_itae:.4f}, OS: {norm_overshoot*100:.2f}%)")

        if np.isnan(loss_wave) or np.isinf(loss_wave):
            print("[Warn] Loss 計算異常，跳過更新。")
            STATE = 1
            continue

        # 重採樣成固定 200 點特徵向量供 1D-CNN 使用
        feature_200 = resample_waveforms(t_arr, target_arr, pos_arr, err_arr, fixed_len=FIXED_LEN)

        target_key = get_target_key(current_target)
        current_best_loss = best_loss_map.get(target_key, 1e9)
        current_best_K = best_K_map.get(target_key, [0.001, 0.0002, 0.5, 0.01])
        noise_scale = current_noise_scale_map[target_key]
        K_old_val = current_K_map[target_key]

        improved = loss_wave < current_best_loss

        if improved:
            noise_scale = max(noise_scale * 0.9, 0.005)
            best_loss_map[target_key] = loss_wave
            best_K_map[target_key] = list(K_old_val)
            print(f"✨ 表現破紀錄！(舊最佳 Loss: {current_best_loss:.6f} -> 新最佳: {loss_wave:.6f})")

            # 存入經驗回放池
            replay_buffer.append((feature_200, list(K_old_val)))
            if len(replay_buffer) > MAX_REPLAY_SIZE:
                replay_buffer.pop(0)

            # 從歷史優良經驗中抽樣訓練 1D-CNN
            batch_size = min(8, len(replay_buffer))
            samples = random.sample(replay_buffer, batch_size)
            batch_features = np.array([s[0] for s in samples], dtype=np.float32)
            batch_targets = np.array([s[1] for s in samples], dtype=np.float32)

            with tf.GradientTape() as tape:
                pred_K = model(batch_features, training=True)
                loss_nn = tf.reduce_mean(tf.square(pred_K - batch_targets))
            grads = tape.gradient(loss_nn, model.trainable_variables)
            optimizer.apply_gradients(zip(grads, model.trainable_variables))
            print(f"  [1D-CNN 訓練完成] Batch Size: {batch_size}, NN Loss: {float(loss_nn):.6f}")

        else:
            noise_scale = min(noise_scale * 1.1, 0.1)
            print(f"未改善表現 (當前最佳 Loss: {current_best_loss:.6f})")

        # 使用 1D-CNN 針對當前波形預測推薦 K 值
        feature_batch = np.expand_dims(feature_200, axis=0) # (1, 200, 4)
        K_pred = model(feature_batch, training=False).numpy()[0]
        K_pred = np.clip(K_pred, [0.0001, 0.00001, 0.01, 0.0001], [2.0, 1.0, 5.0, 1.0])

        # 隨機探索雜訊 (自適應常態分佈)
        noise = np.random.normal(0, noise_scale, size=4)
        noise = np.clip(noise, -0.05, 0.05)
        K_candidate = K_pred + noise

        # 平滑過渡 (Exponential Smoothing)
        alpha = 0.7
        K_old_arr = np.array(K_old_val, dtype=np.float32)
        K_new = alpha * K_old_arr + (1.0 - alpha) * K_candidate
        K_new = np.clip(K_new, [0.0001, 0.00001, 0.01, 0.0001], [2.0, 1.0, 5.0, 1.0])

        # 若連續嚴重惡化 (Loss 超過最佳值 1.3 倍)，觸發 Rollback 保障系統穩定
        if (not improved) and (loss_wave > current_best_loss * 1.3):
            print("⚠️ 觸發 Rollback：重設回歷史最佳 K 參數並縮減探索半徑。")
            K_new = np.array(best_K_map[target_key], dtype=np.float32)
            noise_scale = max(noise_scale * 0.8, 0.005)

        # 更新下一輪參數
        current_K_map[target_key] = K_new.tolist()
        current_noise_scale_map[target_key] = noise_scale

        print(f"下一輪 K 參數: {[round(k, 5) for k in K_new]}")
        print(f"當前探索半徑 Noise Scale: {noise_scale:.4f}")
        print("------------------------------------------------------------------")

        STATE = 1
```
